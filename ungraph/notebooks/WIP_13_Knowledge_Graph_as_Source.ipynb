## Objetivo

Documentar cómo **Ungraph** podría **ingerir desde un grafo de conocimiento existente** (Neo4j, RDF export, otra instancia) como **fuente no estructurada estructurada**:

- Los nodos y relaciones ya son datos; el desafío es **normalizarlos** al patrón `GraphPattern` de destino (p. ej. `FILE_PAGE_CHUNK` o dominio propio).
- Casos de uso: migración entre instancias, **enriquecimiento** de un corpus con triples de un KG externo, o **fusionar** evidencia de grafo A en grafo B con trazabilidad (`PROV-O`, `source_graph_uri`).

## Enfoques posibles

| Enfoque | Descripción |
|---------|-------------|
| **Export → texto** | Cypher `APOC.export` / `CALL` serializa subconjuntos a JSONL o Markdown; luego `ingest_document` sobre archivos generados. |
| **Driver de lectura** | `KnowledgeGraphSourceService` (ABC) que lee nodos `(:Document {text})` y crea `Chunk` sin pasar por PDF/HTML. |
| **SPARQL / RDF** | Si el origen es RDF, convertir a texto por plantilla o usar **GraphRAG** intermedio; fuera del núcleo mínimo. |
