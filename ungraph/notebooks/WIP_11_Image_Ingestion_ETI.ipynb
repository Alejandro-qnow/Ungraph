## Visión del flujo ETI para imágenes

```
Imagen (png/jpeg/webp)
    → Extract: bytes + EXIF + source_id
    → Transform: texto derivado (caption / OCR / JSON estructurado)
    → mismo pipeline que texto: Chunk + embeddings + patrón de grafo
    → Infer: opcional (NER/LLM) sobre el texto derivado
```

## Integración con RAG

- Los **chunks** representan el texto inferido; `metadata` puede incluir `image_sha256`, `bbox` de OCR, o URI del asset.
- Para **evidencia**: relación en grafo entre `Chunk` y nodo `Image`/`Asset` (diseño de patrón pendiente).
