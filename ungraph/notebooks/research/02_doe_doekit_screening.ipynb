**Track:** `docs/experiment/RESEARCH_TRACK.md` · **Oleada 2**

Orquestación **solo** vía `scripts/run_domain_pipeline.py`.
Empezar en `--mode offline`; pasar a `online` con Neo4j + wipe.

Extra: `ungraph[experiments]`.

```python
from pathlib import Path
import subprocess
import sys

REPO = Path("../../..").resolve()
SCRIPT = REPO / "scripts/run_domain_pipeline.py"
DOMAIN = "knowledge_graphs"
MODE = "offline"  # cambiar a "online" cuando Neo4j+spaCy listos

def run_doe(*args: str) -> None:
    cmd = [sys.executable, str(SCRIPT), "--domain", DOMAIN, *args]
    print("$", " ".join(cmd))
    subprocess.check_call(cmd, cwd=REPO)

print("REPO", REPO)
```

## Screening (+ redesign opcional)

```python
run_doe("--design", "screening", "--mode", MODE, "--redesign")
```

## Run celdas del diseño

```python
run_doe("--design", "run", "--mode", MODE)
```

## Analyze (doekit)

```python
run_doe("--design", "analyze")
reports = REPO / "benchmarks/domains/knowledge_graphs/reports"
for name in ("design.json", "results.csv", "summary.json", "analysis.json"):
    p = reports / name
    print(name, "OK" if p.is_file() else "missing", p)
```
